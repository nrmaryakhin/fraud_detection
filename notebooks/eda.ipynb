### В датасете обнаружено 1081 повторное вхождение полных дубликатов. При анализе было выявлено, что некоторые транзакции полностью повторяются несколько раз. При этом наличие одинакового значения Time не является признаком дубликата, поскольку несколько транзакций могли произойти в один и тот же момент. Перед обучением модели полные дубликаты целесообразно удалить, чтобы одинаковые наблюдения не могли одновременно попасть в обучающую и тестовую выборки.

```python
data = data.drop_duplicates()
```

## Визуализация

```python
data["Class"].value_counts(normalize=True) * 100
```

```python
counts = data["Class"].value_counts().sort_index()
percentages = data["Class"].value_counts(normalize=True).sort_index() * 100

ax = counts.plot(kind="bar", figsize=(7, 5))

plt.title("Распределение классов")
plt.xlabel("Class")
plt.ylabel("Количество транзакций")
plt.xticks(rotation=0)

for i, (count, percentage) in enumerate(zip(counts, percentages)):
    ax.text(
        i,
        count + 2000,
        f"{count:,}\n({percentage:.2f}%)",
        ha="center"
    )

plt.show()
```

```python
data.boxplot(column="Amount", by="Class")

plt.title("Распределение суммы транзакций")
plt.suptitle("")
plt.xlabel("Class")
plt.ylabel("Amount")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.scatter(
    data["Time"],
    data["Amount"],
    c=data["Class"],
    alpha=0.3
)

plt.title("Транзакции во времени")
plt.xlabel("Time")
plt.ylabel("Amount")
plt.show()
```

```python
features = data.drop('Class', axis=1).columns.tolist()

for feature in features:
    data.boxplot(column=feature, by="Class")
    plt.title(f"{feature}: распределение по классам")
    plt.suptitle("")
    plt.xlabel("Class")
    plt.ylabel(feature)
    plt.show()
```

```python
corr = data.corr(numeric_only=True)

plt.figure(figsize=(14, 10))
plt.imshow(corr, aspect="auto")
plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Корреляционная матрица")
plt.tight_layout()
plt.show()
```

### Вывод по результатам EDA

В ходе первичного анализа данных было установлено, что в датасете отсутствуют пропущенные значения. Были обнаружены и удалены полные дубликаты записей, после чего данные были подготовлены для дальнейшего моделирования.

Главной особенностью датасета является сильный дисбаланс целевого класса `Class`: мошеннических транзакций значительно меньше, чем обычных. Поэтому использование одной только accuracy для оценки модели будет некорректным — в дальнейшем необходимо ориентироваться на precision, recall, F1-score, ROC-AUC и PR-AUC.

Сравнение признаков между классами с помощью box plot позволяет заметить различия в распределениях некоторых признаков для обычных и мошеннических транзакций. Это говорит о том, что исходные признаки могут содержать информацию, позволяющую разделять классы.
